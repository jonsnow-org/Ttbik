# Kaggle Resume Orchestrator (CPU-only)

**One-time setup, before scheduling this:**
1. Accelerator: **None**. Internet: **On**.
2. Secrets: GITHUB_TOKEN, KAGGLE_USERNAME, KAGGLE_KEY.
3. Edit TARGET_KERNEL_SLUG below to your real training kernel's slug.
4. Save, then Kaggle's own "Schedule this notebook" (every 9 hours).

```python
# قائمة بجميع دفاتر تدريب شام والـ Pipelines المجدولة
TARGET_KERNELS = [
    # --- دفاتر التدريب والـ Tokenizers السابقة ---
    "jonsnowjonsnow/notebookb85993aa78",  # Audio Tokenizer Track
    "jonsnowjonsnow/notebook37c7469512",  # Autonomous Research Track (CPU)
    "jonsnowjonsnow/notebookee02530b8d",  # CPU Training Track
    "jonsnowjonsnow/notebook2d0e40c1f1",  # GPU Training Track
    
    # --- الدفاتر الجديدة المضافة اليوم ---
    "jonsnowjonsnow/notebook04b758ffb5",  # Image Tokenizer Generator Track
    "jonsnowjonsnow/notebookff914258b5",  # Sham Video Corpus Pipeline Track
]
```

```python
import os
import subprocess
from kaggle_secrets import UserSecretsClient

GITHUB_TOKEN = UserSecretsClient().get_secret("GITHUB_TOKEN")
REPO_URL = f"https://{GITHUB_TOKEN}@github.com/jonsnow-org/Ttbik.git"
BRANCH = "claude/free-services-marketplace-h6rwk2"
CLONE_DIR = "/kaggle/working/Ttbik"

if not os.path.exists(CLONE_DIR):
    subprocess.run(["git", "clone", "--depth", "1", "--branch", BRANCH, REPO_URL, CLONE_DIR], check=True)
else:
    subprocess.run(["git", "-C", CLONE_DIR, "pull"], check=True)

subprocess.run(["pip", "install", "-q", "-U", "kaggle"], check=True)
print("repo ready, kaggle package installed.")
```

```python
import os
import sys
from kaggle_secrets import UserSecretsClient

KAGGLE_USERNAME = UserSecretsClient().get_secret("KAGGLE_USERNAME")
KAGGLE_KEY = UserSecretsClient().get_secret("KAGGLE_KEY")
os.environ["KAGGLE_USERNAME"] = KAGGLE_USERNAME
os.environ["KAGGLE_KEY"] = KAGGLE_KEY

sys.path.insert(0, "/kaggle/working/Ttbik/ai-system/scripts")
from kaggle_auto_resume import resume_kernel

# تشغيل كل دفتر من دفاتر التدريب الأربعة
for target_slug in TARGET_KERNELS:
    print(f"\n==========================================")
    print(f"Triggering resume for: {target_slug}")
    print(f"==========================================")
    
    try:
        result = resume_kernel(
            kernel_slug=target_slug,
            notebook_repo_path="/kaggle/working/Ttbik/ai-system/colab/sham_small/kaggle_notebooks",
            work_dir="/kaggle/working/kaggle-kernel",
            sync_from_repo=False,
        )
        print(f"Result for {target_slug}:\n{result}")
    except Exception as e:
        print(f"Error resuming {target_slug}: {e}")

print("\n--- Completed orchestration for all notebooks ---")
```
